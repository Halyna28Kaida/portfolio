```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url = "https://www.scrapethissite.com/pages/forms"
```

```python
main_page = requests.get(url)
page = "/?page_num="
```

```python
soup = BeautifulSoup(main_page.text, "html")
soup
```

```python
table_head = soup.find_all("th")
heads = [head.text.strip() for head in table_head]
print(heads)
```

```python
df = pd.DataFrame(columns=heads)
df
```

```python

pages = [a.get_text(strip=True) for a in soup.find_all('a') if a.get_text(strip=True).isdigit()]
```

```python
def create_rows_data(soup):
  row_data = soup.find_all("tr", class_="team")
  for row in row_data:
    td_row = row.find_all("td")
    ind_row = [td.text.strip() for td in td_row]
    length = len(df)
    df.loc[length] = ind_row
```

```python
for num in pages:
  page_url= requests.get(f'{url}{page}{num}')
  soup = BeautifulSoup(page_url.text, "html")
  create_rows_data(soup)
df
```
